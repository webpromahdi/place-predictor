```python


import joblib
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')   # Non-interactive backend (safe for notebooks + servers)

# ------------------------------------------------------------------
# 1. Build comparison table
# ------------------------------------------------------------------
comparison_data = []
#---model_results = {}  # This should already be populated by all 6 members before running this cell
for model_name, metrics in model_results.items():
    comparison_data.append({
        'Model':     model_name,
        'Accuracy':  round(metrics['accuracy'],  4),
        'Precision': round(metrics['precision'], 4),
        'Recall':    round(metrics['recall'],    4),
        'F1-Score':  round(metrics['f1'],        4)
    })

comparison_df = pd.DataFrame(comparison_data)
comparison_df = comparison_df.sort_values('F1-Score', ascending=False).reset_index(drop=True)
comparison_df.index += 1  # Start rank from 1, not 0

print("=" * 65)
print("MODEL COMPARISON TABLE (sorted by F1-Score — our primary metric)")
print("=" * 65)
print(comparison_df.to_string())
print()

# ------------------------------------------------------------------
# 2. Select the champion (highest F1-Score)
# ------------------------------------------------------------------
best_model_name = comparison_df.iloc[0]['Model']   # First row = highest F1
best_f1         = comparison_df.iloc[0]['F1-Score']
champion_pipeline = model_results[best_model_name]['pipeline']

print(f"Champion model  : {best_model_name}")
print(f"Champion F1     : {best_f1:.4f}")
print()

# ------------------------------------------------------------------
# 3. Export as champion_model.pkl
# ------------------------------------------------------------------
# os.makedirs('models', exist_ok=True)
champion_path = '../models/champion_model.pkl'
joblib.dump(champion_pipeline, champion_path)

print(f"Champion model saved → {champion_path}")
print("The Streamlit app (src/app.py) loads this file directly.")
print()

# ------------------------------------------------------------------
# 4. Save the comparison table as a chart (used in Streamlit Tab 2)
# ------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 4))
ax.axis('off')

table = ax.table(
    cellText=comparison_df.values,
    colLabels=comparison_df.columns,
    loc='center',
    cellLoc='center'
)
table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1.2, 1.8)

# Highlight the champion row (row 1 = index 1 in table = data row 0)
for col_idx in range(len(comparison_df.columns)):
    table[(1, col_idx)].set_facecolor('#d4edda')   # Light green highlight
    table[(1, col_idx)].set_text_props(weight='bold')

# Style header row
for col_idx in range(len(comparison_df.columns)):
    table[(0, col_idx)].set_facecolor('#343a40')
    table[(0, col_idx)].set_text_props(color='white', weight='bold')

plt.title(f'Model Comparison — Champion: {best_model_name} (F1={best_f1:.4f})',
          fontsize=13, pad=20, weight='bold')
plt.tight_layout()

chart_path ='../charts/model_comparison_table.png'
plt.savefig(chart_path, dpi=150, bbox_inches='tight')
plt.close()
print(f"Comparison chart saved → {chart_path}")
```
